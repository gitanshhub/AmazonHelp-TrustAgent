# Milestone 1 & 2: Dataset Exploration & Brand Selection

This notebook inspects the **Customer Support on Twitter** dataset (`twcs/twcs.csv`), computes dataset-wide statistics, analyzes brand support volumes, and provides empirical justification for selecting the target brand.

```python
import os
import json
import pandas as pd

summary_path = "../data/brand_ranking.json"
if not os.path.exists(summary_path):
    summary_path = "data/brand_ranking.json"

with open(summary_path, "r", encoding="utf-8") as f:
    data = json.load(f)

print(f"Total Tweets: {data['total_tweets']:,}")
print(f"Customer Messages (Inbound): {data['inbound_count']:,} ({data['inbound_count']/data['total_tweets']*100:.1f}%)")
print(f"Brand Messages (Outbound): {data['outbound_count']:,} ({data['outbound_count']/data['total_tweets']*100:.1f}%)")
print(f"Unique Brands: {data['unique_brands']:,}")
```

## Top Brands Ranked by Volume and Multi-Turn Depth

```python
df_brands = pd.DataFrame(data['brands'])
df_brands['multi_turn_ratio'] = (df_brands['multiturn_replies'] / df_brands['total_tweets']).round(3)
df_brands
```

## Brand Selection Justification

> **Selected Brand: AmazonHelp**
>
> **Empirical Justification**:
> 1. **Highest Support Volume**: `AmazonHelp` accounts for **169,840 brand support tweets**, the highest across all 108 brands.
> 2. **Densest Multi-turn Conversations**: With **85,274 multi-turn interactions** (over 50.2% multi-turn reply rate), AmazonHelp provides rich dialog trees with back-and-forth problem diagnostics and resolutions.
> 3. **Diverse & Actionable Support Intents**: Covers rich e-commerce intents (late deliveries, missing parcels, refund requests, return authorization, Prime billing, damaged items, payment disputes).
> 4. **Clear Trust / Safety Boundaries**: Distinguishes between safely auto-handleable inquiries (delivery tracking policies, return windows, self-serve link directions) and mandatory human escalation (card theft, missing high-value shipments, account takeover).
